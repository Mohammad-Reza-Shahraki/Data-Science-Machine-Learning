```python
arr = [1, 2, 15, 9, 6, 5]
arr = numpy.array(arr)
arr
```

```python
# len = تعداد عناصر
# // = قسمت صحیح تقسیم
len(arr) // 2
```

```python
p = arr[len(arr)//2]
p
```

```python
l = [x for x in arr if x < p]
# l = numpy.array(l)
l
# این یک لیست پایتونی می دهد اما اگر دستور خط دوم را می نوشتیم، یک لیست نامپایی به ما تحویل می داد
```

```python
m = [x for x in arr if x > p]
m
```

```python
n = [x for x in arr if x == p]
n
```

```python
l + m + n
```

```python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    arr = numpy.array(arr)
    p = arr[len(arr)//2]
    l = [x for x in arr if x < p]
    m = [x for x in arr if x == p]
    n = [x for x in arr if x > p]
    return quicksort(l) + m + quicksort(n)
```

```python
quicksort(arr)
```

```python
h = [3, 6, 8, 9, 15, 27, 5, 1]
quicksort(h)
```
