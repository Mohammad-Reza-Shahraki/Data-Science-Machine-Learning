```python
m1 = numpy.zeros((3, 4), int)
m1
```

```python
m2 = numpy.ones((3, 4), int)
m2
```

```python
numpy.vstack([m1, m2])
# V : VERTICAL عمودی
# برای این دستور باید تعداد مولفه های دوم(ستون ها) در هر لیست برابر باشد
```

```python
numpy.hstack([m1, m2])
# H : HORIZONTLE افقی
# برای این دستور باید تعداد مولفه های اول(سطر ها) در هر لیست برابر باشد
```

```python
# برای اجرای دو دستور بالا، ترتیب ورودی ها در ترتیب خروجی موثر است
```

```python
nyek.T
# T = TRANSE جای سطر ها را با ستون ها عوض میکند
# We should write T NOT t
# ndo = nyek.T این دستور یک متغیر دیگر تعرف نمیکند و به متغیر دیگر دست نمیزند گویی این دستور را وارد کرده ایم
```

```python
nyek.astype('f')
# تایپ را از حالت اینتیجر(عدد صحیح) به فلوت(عدداعشاری) در یک متغیر دیگر تغییر میدهد
```

```python
nyek.sum()
# جمع مقادیر عناصر جداول
```

```python
nyek.max()
# بزرگترین مقدار جداول را نشان میدهد
```

```python
nyek.min()
# کوچکترین مقدار جداول را نشان میدهد
```

```python
nyek.argmax()
# جایگاه(ایندکس) بزرگترین عدد را نشان میدهد که چندمین عنصر ماتریس است(فرایند شمارش از صفر شروع میشود)
```

```python
nyek.argmin()
# جایگاه(ایندکس) کوچکترین عدد را نشان میدهد که چندمین عنصر ماتریس است(فرایند شمارش از صفر شروع میشود)
```

```python
nyek.mean()
# میانگین عناصر
```

```python
nyek.std()
# انحراف استاندارد عناصر
```

```python
mr = numpy.arange(15) ** 2
# یعنی از رنج صفر تا چهارده(پانزده عدد) یک آرایه ی نامپایی بساز و در آخر به توان دو کن
mr
```

```python
mr[2]
# یعنی ایندکس دو را به من نشان بده
```

```python
mr[2 : 6]
# یعنی ایندکس دو تا شش را به من نشان بده
```

```python
mr[ : 6]
# یعنی ایندکس هیچی(اول) تا شش را به من نشان بده
```

```python
mr[6 : ]
# یعنی ایندکس شش تا هیچی(آخر) را به من نشان بده
```

```python
mr[-5 : ]
# یعنی ایندکس از آخر پنجی تا آخر را به من نشان بده
```

```python
mr[ : -2]
# یعنی از هیچی(اول) تا از آخر دومی را به من نشان بده
```

```python
mr[4 : 13 : 2]
# از ایندکس چهار تا سیزده، دوتا دوتا برو جلو و انتخاب کن و به من نشان بده
```

```python
mr[4 : : 2] 
#  or mr[4::2]
# یعنی از ایندکس چهار تا هیچی(آخر) دوتا دوتا برو جلو و انتخاب کن و به من نشان بده
```

```python
mr[4 : : -2]
# یعنی از ایندکس چهار تا هیچی(چون قدم برداری معکوسه، پایان این دستور همان اول است) دوتا دوتا برو عقب و انتخاب کن و به من نشان بده
```

```python
mr[-10 : -2 : 2]
# یعنی از ایندکس از آخر دهمی تا از آز آخر دومی دوتا دوتا برو جلو و به من نشان بده
```
